```python
from sklearn.neighbors import LocalOutlierFactor
from sklearn.metrics import confusion_matrix, classification_report
import pandas as pd

# Charger le dataset depuis l'environnement précédent
df = pd.read_csv("/content/drive/MyDrive/ColabNotebooks/detection_anomalie_transaction_bancaire/creditcard.csv")
df = df.drop_duplicates()

# Préparation des données
X = df.drop('Class', axis=1)
y = df['Class']

# Appliquer le modèle LOF (Local Outlier Factor)
# Important : LOF ne propose pas de méthode .predict() sur données séparées
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.003, novelty=False)
y_pred = lof.fit_predict(X)

# Mapping : LOF retourne -1 = anomalie → on remappe
df['anomaly_pred'] = pd.Series(y_pred, index=df.index).map({1: 0, -1: 1})

# Évaluation du modèle
cm = confusion_matrix(y, df['anomaly_pred'])
report = classification_report(y, df['anomaly_pred'], output_dict=True)
report_df = pd.DataFrame(report).transpose()

# Extraire et calculer les indicateurs clés
TN, FP, FN, TP = cm.ravel()
recall = TP / (TP + FN)
precision = TP / (TP + FP)

f1 = 2 * (precision * recall) / (precision + recall)

# Résumé
summary = {
    "True Positives": TP,
    "False Positives": FP,
    "False Negatives": FN,
    "True Negatives": TN,
    "Recall": round(recall, 4),
    "Precision": round(precision, 4),
    "F1-score": round(f1, 4)
}

print('Summary : \n', summary)
```
